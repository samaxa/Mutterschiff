Um ein Netzwerk aufzubauen nutzen wir den Datentyp Network aus PyPSA. Hierzu initialisieren wir eine Variable mit dem Befehel "network_1 = pypsa.Network()". Diese Variable ist unser "Container" in dem alle weiteren Komponenten hinzugefügt werden. <br>
Desweiteren müssen wir die Anzahl an Snapshots feslegen. Dies sind in unserem Fall die 365 Tage. Dafür nutzen wir den Befehel "network_1.set_snapshots(range(0,365))".

Mit der Methode .add() können Komponenten dem Netzwerk hinzugefügt werden. Die wichtigsten sind dabei
* Bus
* Generator
* Load
* Line
* Link
* Store
* Storage Unit <br>
Welche Eigenschaften die jeweiligen Komponenten haben und welche weiteren es noch gibt kann man unter folgendem Link gut nachvollziehen. <br>
https://docs.pypsa.org/v0.35.1/user-guide/components.html

Nun da das Netzwerk aufgebaut ist, können wir es optimieren lassen. Dies geschieht über den Befehl "network_1.optimize(solver_name = 'gurobi')"
