### Aufgabe 1: Bedingte Anweisungen

Eine Windkraftanlage benötigt für den Betrieb eine Windstärke von mindestens 13 km/h. Ab 50 km/h erreicht Sie ihre maximale Leistung. Dazwischen steigt die Leistung linear an. Im Falle eines Sturms muss die Anlage ab einer Windstärke von 90 km/h abgeschalten werden, um Schäden zu vermeiden. Erzeugen Sie sich eine zufällige Zahl zwischen 0 und 120. Dies soll die Windgeschwindigkeit darstellen. Daraufhin geben Sie eine Nachricht aus, bei welcher relativen Leistung die Anlage läuft, bzw. ob diese abgeschalten ist. Nutzen Sie hierbei bedingte Anweisungen. <br>
Für die Erzeugung der Zufallszahl benötigen Sie die Bibliothek "random". Nutzen hieraus die Funktion "random.uniform(start,end)". Diese Erzeugt eine zufällige Gleitkommazahl (float) innerhlab der gegebenen Grenzen.

```python
import random
```

```python
wind = random.uniform(0,120)
wind
```

```python
wind = random.uniform(0,120)
if wind < 13:
    status = "Anlage ist abgeschaltet"
elif wind > 90:
    status = "Anlage ist abgeschaltet"
elif wind > 50:
    status = "Anlage bei 100% Leistung"
else:
    leistung = round((wind-13) / (50-13) * 100, 2)
    status = f"Anlage bei {leistung}% Leistung"       

print(f"Windgeschwindigkeit: {round(wind,1)} km/h.", status)
```

### Aufgabe 2: Funktionen

Es soll eine Solarstrahlungsangebotsverlauf synthetisiert werden. Dieser kann näherungsweise durch eine trigonometrische Funktion beschrieben werden. Zusätzlich wird die Einstrahlungsmenge benötigt. <br> 
In Deutschland trifft im Sommer ca. 5,5 kWh/Tag Solarstrahlung horizontal auf die Erdoberfläche. <br>
Synthetisiere ein Solarstrahlungsverlauf für ein Jahr in Deutschland mit einer täglichen Auflösung.

Importieren Sie znächst die Bibliothek "math"

```python
import math
```

Sie benötigen die Funktion "math.sin()" um mit den Sinus rechnen zu können.
Nutzen Sie den Ausdruck "math.sin(day * (2 * math.pi / 730))" um die Funktion für die Solarstrahlung in Abhängigkeit des Tages aufzustellen.

```python
day = 120
max_radiation = math.sin(day * (2 * math.pi / 730)) * 5.5 #kWh/m^2
print(round(max_radiation,2), "kWh/m^2")
```

Fügen Sie nun der Formel eine zufällige Verschattung zwischen 0 und 100% hinzu. Nutzen Sie hierbei die bereits kennengelernte Funktion "random.uniform()". <br>
Beachten Sie, dass die Einstrahlung nicht negativ werden kann!

```python
day = 120
max_radiation = (math.sin(day * (2 * math.pi / 730)) - random.uniform(0,1)) * 5.5 #kWh/m^2
if max_radiation < 0:
    max_radiation = 0
print(round(max_radiation,2), "kWh/m^2")
```

Nutzen Sie den von Ihnen erstellten Ausdruck um eine Funktion zu schreiben, die Ihnen die Solarstrahlung für den angegeben Tag ausgibt.

```python
def syn_max_radiation(day):
    max_radiation = (math.sin(day * (2 * math.pi / 730)) - random.uniform(0,1)) * 5.5 #kWh/m^2
    if max_radiation < 0:
        max_radiation = 0
    return max_radiation
```

### Aufgabe 3: Schleifen

Nutzen Sie die erstellte Funktion um den Verlauf für ein ganzes Jahr zu synthetisieren. Nutzen Sie hierfür eine Schleife. <br>
Tipp: Sie benötigen die methode 'range(start, stop)'

```python
liste_solar = []
for tag in range(0, 365):
    berechnung = syn_max_radiation(tag)
    liste_solar.append(round(berechnung, 2))

print(liste_solar)
```

Wir benötigen später einen normierten Verlauf, sprich einen Verlauf zwischen 0 und 1 (100%). Hierzu können wir den Datentyp "array" von Numpy nutzen. Dieser erlaubt es uns mathematische Operationen über alle Elemente der Liste durchzuführen.

```python
array_solar = np.array(liste_solar)
array_solar_min_max = array_solar/array_solar.max()
# ds = pd.Series(array_solar_min_max)
# ds.plot()
```

Wir brauchen neben dem Solarstrahlungsangebot auch den Verlauf der Windstärke. Dieser wird nun der Einfachheit halber als eine Normalverteilung der Windgeschwindigkeiten mit einem Mittelwert von 20 km/h und einer Standardabweichung von 20 angenommen. <br>
Verwenden Sie die Funktion aus der Numpy Bibliothek "np.random.normal(loc=20, scale=20, size=365)". Diese erzeugt ein Numpy array mit 365 Elementen mit zufälligen Werten, welche normalverteil sind.

```python
windstärke = np.random.normal(loc=20, scale=20, size=365)
```

```python
windstärke
```

Nun müssen - wie in Aufgabe 1 - die Wundgeschwindigkeiten der resultierenden Leistung zugeordnet werden.

```python
idx = 0
for wind in windstärke:
    if wind < 13:
        windstärke[idx] = 0
    elif wind > 90:
        windstärke[idx] = 0
    elif wind > 50:
        windstärke[idx] = 100
    else:
        windstärke[idx] = round((wind-13) / (50-13) * 100, 2)
    idx += 1
```

```python
windstärke = windstärke/100
```

Um uns beide Verläufe in einer Grafik darstellen lassen zu können, müssen wir ein Pandas Dataframe erstellen. Hierzu nutzen wir zunächst die Funktion zip() um eine Reihe an Wertepaarungen zu erhalten. 

```python
data = zip(array_solar_min_max, windstärke)
```

Erstellen Sie nun ein Dataframe mit der Funktion "pd.DataFrame()". Die Daten werden mit dem Argument "data = ", die Namen der Spalten mit "columns = " und der Index mit "index = " eingelesen.

```python
df_pv_wind = pd.DataFrame(data = data,columns = ['PV', 'Wind'])
df_pv_wind
```
