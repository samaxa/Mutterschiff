### Aufgabe 2: Methoden für Listen

Im folgenden haben wir den Wocheneinkauf eines typischen Studis in Form eines Textes.

```python
einkaufzettel = "2 x Nudeln je 1,5€ \n 4 x Pesto je 3€ \n 5 x Mate je 0,89€"
print(einkaufzettel)
```

Erstelle eine Liste mit den jeweiigen Produkt, dessen Anzahl und Preis. <br>
Tipp: Durch die Methode "split()" wird eine Liste erzeugt.

```python
einkaufzettel_list = einkaufzettel.replace(' ','').splitlines()
einkaufzettel_list
```

Jedes Element der Liste soll nun widerum eine eigene Liste sein, sodass Anzahl, Produkt und Preis ein eigenes Objekt sind.

```python
# einkaufzettel_list[0] = einkaufzettel_list[0].replace('je','x').split('x')
# einkaufzettel_list[1] = einkaufzettel_list[1].replace('je','x').split('x')
# einkaufzettel_list[2] = einkaufzettel_list[2].replace('je','x').split('x')
print(einkaufzettel_list)
idx = 0
for produkt in einkaufzettel_list:
    einkaufzettel_list[idx] = produkt.replace('je','x').split('x')
    idx +=1

einkaufzettel_list
```

Füge für jedes Produkt die Kosten für dieses der Sub-Liste hinzu. (Anzahl * Kosten)

```python
einkaufzettel_list[0].append(float(einkaufzettel_list[0][0]) * float(einkaufzettel_list[0][2].replace(",",".").replace('€','')))
einkaufzettel_list[1].append(float(einkaufzettel_list[1][0]) * float(einkaufzettel_list[1][2].replace(",",".")[:-1]))
einkaufzettel_list[2].append(float(einkaufzettel_list[2][0]) * float(einkaufzettel_list[2][2].replace(",",".")[:-1]))
```

```python
einkaufzettel_list
```

```python
einkaufzettel_list[0][-1]
```

Wie teuer ist der gesamte Einkauf?

```python
ges_kosten = einkaufzettel_list[0][3] + einkaufzettel_list[1][-1] + einkaufzettel_list[2][-1]
print(ges_kosten, "€")
```

### Aufgabe 3: Dynamische Strings

Gegeben sind das Jahr, die Leistung und die Volllaststunden einer PV-Anlage.

```python
year = 2020
pv_power_kW = 0.8
pv_full_load_hours = 1001
print(year,";", pv_power_kW , "kW ;", pv_full_load_hours , "h")
```

Da Strings unveränderliche Datentypen sind muss man einen neuen String erstellen um die alten zu verändern. <br>
Alternativ gibt es dynamische Lösungen unter Verwendung von f-Strings, .format()-Methode, oder .replace()-Methode. <br>
Bauen Sie mit allen drei Varianten einen String auf um folgenden Satz zu realisieren:
"Jahr: 2020; Leistung: 0,8 kW; Volllaststunden: 1001 h"

Verwendung von f-Strings:

```python
pv_datasheet = f"Jahr: {year}; Leistung: {pv_power_kW} kW; Volllaststunden: {pv_full_load_hours} h"
print(pv_datasheet)
```

Verwendung von .format():

```python
pv_datasheet = "Jahr: {}; Leistung: {} kW; Volllaststunden: {} h"
pv_datasheet = pv_datasheet.format(year, pv_power_kW, pv_full_load_hours)
print(pv_datasheet)
```

Verwednung von .replace():

```python
pv_datasheet_mask = "Jahr: year; Leistung: power kW; Volllaststunden: hours h"
pv_datasheet = pv_datasheet_mask.replace("year", str(year)).replace("power", str(pv_power_kW)).replace("hours", str(pv_full_load_hours))
print(pv_datasheet)
```
