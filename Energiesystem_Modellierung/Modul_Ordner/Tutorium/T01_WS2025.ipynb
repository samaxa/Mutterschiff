### Aufgabe 1: Variablen, Zuweisungen und Arithmetische Operatoren

Ein Einfamilienhaus hat eine 7 kWp PV-Anlage auf dem Dach und erzeugte mit dieser im Jahr 2022 ca. 9 MWh elektrische Energie.
Erstellen Sie Variablen mit aussagekräftigen Namen und weisen Sie diesen die entsprechende Werte für Peakleistung und Energiemenge zu.

```python
pv_max_power_kW = 7
pv_yearly_energy_generation_MWh = 9
```

Wie viel Leistung weist die PV-Anlage im Schnitt auf? Gebe das Ergebnis auf zwei Nachkommastelle genau an. <br>
Tipp: Sie benötigen die Anzahl an Stunden in einem Jahr. <br>
Tipp: Mit der Funktion "round(Variable,anz_nachkommastellen)" wird der Wert der Variable gerundet.

```python
pv_avg_power_kW = round(pv_yearly_energy_generation_MWh * 1000 / 8760,2)
print(pv_avg_power_kW,"kW")
```

Wie vielen Volllaststunden entspricht dies? Gebe das Ergebnis als Integer in Stunden an.

```python
pv_full_load_hours = int(pv_avg_power_kW/pv_max_power_kW * 8760)
print(pv_full_load_hours, "h")
print(type(pv_full_load_hours))
```

```python
pv_yearly_energy_generation_MWh += 3
```

### Aufgabe 2: Datentypen und Datenkonvertierung

Im folgenden sind zwei Variablen gegeben.

```python
var_1 = 5
var_2 = 2.5
```

Wie erhält man durch die Konvertierung der Datentypen die Werte: 7.5; 7; 10 und 52.5? <br>
(jede Variable soll genau einmal Verwendet werden um den Wert zu erhalten) <br>
Tipp: Sie benötigen nur Multiplikation, Addition und Datenkonvertierung <br>
Was passiert wenn man einen String mit einem Integer multipliziert?

```python
# 7.5 erhalten durch Addition
print(var_1 + var_2)
# 7 erhalten durch Addition und Konvertierung von var_2 in Integer
print(var_1 + int(var_2))
# 10 erhalten durch Multiplikation und Konvertierung von var_2 in Integer
print(var_1 * int(var_2))
# 52.5 erhalten durch Addition und vorheriger Konvertierung von var_1 und var_2 in Strings
print(float(str(var_1)+str(var_2)))
# Ergebnis der Multiplikation eines Strings mit einem Integer
print('Durch die Multiplikation eines Strings mit einem Integer wird der String mit der Anzahl der Multiplikationen hintereinander geschrieben: ',
      int(var_1) * str(var_2))
```

### Aufgabe 3: Logische- und Vergleichsoperatoren

Stell dir vor, du programmierst eine einfache Ampelsteuerung zur Verkehrsregelung.

Die Regeln sind:
* Die Ampel soll grün zeigen, wenn mindestens 5 Autos an der Ampel warten.
* Die Ampel soll auch grün zeigen, wenn gerade ein Notfallfahrzeug naht (z.B. Krankenwagen oder Feuerwehr), unabhängig von der Anzahl wartender Autos.

Es befinden sich aktuell 3 Autos an der Ampel und kein Notfallfahrzeug. <br>
Erstelle Variablen die das abbilden.

```python
anzahl_autos = 3
anzahl_notfallfahrzeug = 0
```

Schreibe eine Bedingung, die überprüft, ob genug Autos an der Ampel warten.

```python
genug_autos = anzahl_autos >= 5
genug_autos
```

```python
notfallfahrzeug = bool(anzahl_notfallfahrzeug)
notfallfahrzeug
```


Kombiniere diese beiden Bedingungen, so dass die Ampel grün zeigt, wenn mindestens 5 Autos warten oder ein Notfallfahrzeug naht.

```python
ampel_gruen = genug_autos or notfallfahrzeug
ampel_gruen
```

```python
print("Ampel ist grün:", ampel_gruen)
```
