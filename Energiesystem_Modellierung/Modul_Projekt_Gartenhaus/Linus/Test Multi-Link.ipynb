```python
import pypsa
import pandas as pd
```

```python
# Netzwerk erstellen
network = pypsa.Network()
network.set_snapshots(range(8760))

#Buses
# Input Busses
network.add('Bus', name = 'I_Heat')
network.add('Bus', name = 'I_Cooling')
network.add('Bus', name = 'I_Light')
network.add('Bus', name = 'I_Water')

# Output Busses
network.add('Bus', name = 'O_Calories')
network.add('Bus', name = 'O_Fats')
network.add('Bus', name = 'O_Proteins')
network.add('Bus', name = 'O_Fiber')


network.add("Link", name="Kartoffel",
# Inputs
    bus0="I_Water",

# Indirect Inputs
    bus1="I_Light",
    bus2="I_Cooling",
    bus3="I_Heat",

# Outputs
    bus4="O_Calories",
    bus5="O_Fats",
    bus6="O_Proteins",
    bus7="O_Fiber",

    p_nom=1.0,

    efficiency=-1.0,    # Water --> Cooling (negativ)
    efficiency2=1.0,    # Heat --> Fats
    efficiency3=1.0,    # Heat --> Proteins
    efficiency4=1.0     # Water --> Calorie
)
```
