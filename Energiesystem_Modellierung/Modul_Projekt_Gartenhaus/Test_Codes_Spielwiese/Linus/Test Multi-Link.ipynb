```
import numpy as np
import pypsa
import pandas as pd
'Änderung Linus'
```

```
# Netzwerk erstellen
network = pypsa.Network()
network.set_snapshots(range(8760))

#Buses
# Input Busses
network.add('Bus', name = 'Heat')
network.add('Bus', name = 'Cooling')
network.add('Bus', name = 'Electricity')

network.add('Generator',
            name = 'gen1',
            bus = 'Electricity',
            p_nom = 1000,
            marginal_cost = 1)


network.add('Load',
            name = 'Heat Load',
            bus = 'Heat',
            p_set = 10)

network.add('Load',
            name = 'Cooling Load',
            bus = 'Cooling',
            p_set = 5)

network.add("Link",
            name="WP",
            bus0="Electricity",
            bus1="Heat",
            bus2="Cooling",
            p_nom_extendable = True,
            efficiency=1.0,
            efficiency2=1.0,
)
network.optimize()
```

```
network.links_t.p0.plot()
network.links_t.p1.plot()
network.links_t.p2.plot()
```
